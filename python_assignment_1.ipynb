```python
# 1.
age = int(input("Enter your age: "))
if age < 5:
    print("Your ticket is free")
elif age <= 17:
    print("Your ticket price is Rs. 100")
elif age <= 59:
    print("Your ticket price is Rs. 200")
else:
    print("Your ticket price is Rs. 120")
```

```python
# 2 

units = int(input("Enter units consumed: "))
if units <= 100:
    bill = units * 3
elif units <= 200:
    bill = (100 * 3) + (units - 100) * 5
else:
    bill = (100 * 3) + (100 * 5) + (units - 200) * 8
print("Total bill: Rs.", bill)
```

```python
#3 

number = int(input("Enter a number: "))
for i in range(1, 11):
    print(number, "x", i, "=", number * i)
```

```python
#4

total = 0
days = 0
while total < 1000:
    savings = int(input("Enter today's savings: "))
    total = total + savings
    days = days + 1
print("Goal reached in", days, "days!")
print("Total saved: Rs.", total)
```

```python
# 5 
correct_pin = 4321
attempts = 0
while attempts < 3:
    pin = int(input("Enter PIN: "))
    if pin == correct_pin:
        print("Access granted")
        break
    else:
        print("Wrong PIN")
        attempts = attempts + 1
if attempts == 3:
    print("Card blocked")
```

```python
#6

N = int(input("Enter N: "))
even = 0
odd = 0
for i in range(1, N + 1):
    if i % 2 == 0:
        even = even + 1
    else:
        odd = odd + 1
print("Even numbers:", even)
print("Odd numbers:", odd)
```

```python
#7
secret = 27
guesses = 0
while True:
    guess = int(input("Guess the number: "))
    guesses = guesses + 1
    if guess == secret:
        print("Correct! You took", guesses, "guesses.")
        break
    elif guess > secret:
        print("Too high")
    else:
        print("Too low")

```

```python
#8
marks = int(input("Enter your marks: "))
if marks < 0 or marks > 100:
    print("Invalid marks")
elif marks >= 90:
    print("Grade A")
elif marks >= 75:
    print("Grade B")
elif marks >= 50:
    print("Grade C")
elif marks >= 35:
    print("Grade D")
else:
    print("Fail")
```

```python
#9
bill = int(input("Enter bill number: "))
total = 0
while bill > 0:
    digit = bill % 10
    total = total + digit
    bill = bill // 10
print("Sum of digits:", total)
if total > 20:
    print("Congratulations! You win a free gift.")
else:
    print("Sorry! No free gift.")
```
